```python


df5 = df.copy()

numeric_cols = df5.select_dtypes(include=['int64','float64']).columns

for col in numeric_cols:

    while True:

        Q1 = df5[col].quantile(0.25)
        Q3 = df5[col].quantile(0.75)
        IQR = Q3 - Q1

        lower = Q1 - 1.5*IQR
        upper = Q3 + 1.5*IQR

        old_rows = len(df5)

        df5 = df5[(df5[col] >= lower) & (df5[col] <= upper)]

        new_rows = len(df5)

        if old_rows == new_rows:
            break

df5.reset_index(drop=True, inplace=True)

print(df5.shape)
```

```python
sns.boxplot(x=df5["medv"])
plt.show()
```
